### Login system using function

```python
import re
data={}    
def verification(Email,Password):
    if re.match(r"^[a-z]+[0-9]*@gmail.com$",Email) and re.match(r"^[A-Z]+[a-z]+[@#$%^&*]+[0-9]+$",Password):
        return True
    else:
        return False
        
def register(F_name,L_name,Email,Password):
    if verification(Email,Password)==True:    
        if Email in data:
            return ( f"{Email} already exist Please Login")
        else:
            data[Email]={"F_name":F_name,"L_name":L_name,"Password":Password}
            #print(data)
            return f"Registration successfull...!"
    else:
        return "enter valid email or password"
        
def login(email,password):
    if verification(email,password)==True:
        if email in data:        
            if data[email]["Password"]==password:
                return f"Logged in\nWell Come {data[email]['F_name']}"
            else:
                return "Either password or email is incorrect"
        else:
            return "Email not registered please register."
    else:
        return "Invalid email or password"
        
      
def control():
    while True:
        enter=int(input("1.Register\n2.Login\n"))
        if enter==1:
            F_name=input("\nEnter your First name:")
            L_name=input("Enter your Last name:")
            Email=input("Enter your Email:")
            Password=input("Enter password")
            print(register(F_name,L_name,Email,Password))
        elif enter==2:
            Email=input("\nEmail:")
            Password=input("Password:")
            print(login(Email,Password))
            for key,value in data.items():
                for k,v in value.items():
                    print( "\ndetails of all users:\nEmail:",key,"\nFirst Name:",k,"\nLast Name:",v)
                    break
        else:
            print("good bye...!")
            break
                    
            
control()
```

### 1.Count even and odd numbers

```python
numbers=[10,15,22,33,48,51]
even=0
odd=0
for i in numbers:
    if i%2==0:
        even+=1
       
    else:
        odd+=1
print(f"count of even numbers:{even}")
print(f"count of odd numbers:{even}")
```

### 2. Find the largestnumber

```python
numbers=[45,78,12,98,23]
def find_largest(numbers):
    large=float('-inf')
    for i in numbers:
        if i>large:
            large=i
    return large
print(find_largest(numbers))
        
```

### 3.Remove duplicate items

```python
item=["Apple","Banana","Apple","Orange","Banana"]
def duplicate(items):
    dup=[]
    for i in items:
        if i not in dup:
            dup=dup+[i]
    return dup
print(duplicate(item))
```

### 5.Shopping cart

```python
items=[]
def add(item):
    if item not in items:
        items.append(item)
    else:
        print(f"{item} already in cart")

def display():
    return items
def main():
    while True:
        
        c=int(input("1.Add item\n2.View cart\n3.Exit\n"))
        if c==1:
            item=input("Enter item to add:")
            add(item)
        elif c==2:
            c=input("Do you wish to display all items (y/n)?:")
            if c.lower()=="y":
                print(f"cart items:{display()}")
            else:
                break
        elif c==3:
            break
main()
```

### 7.Count Word Frequency

### 9.Inventory Management

```python
inventory=[]
def add_product(product):
    if product in inventory:
        print( f"Product Already Exists\n")
    else:
        inventory.append(product)
        print( f"{product} added successfully\n")
        
def search(product):
    if product in inventory:
        return f"{product} is found\n"
    else:
        return "product not found\n"
        
def remove(product):
    if product in inventory:
        inventory.remove(product)
        return f"{product} removed  successfully\n"
    else:
        return "product not found\n"
        
def display():
    return inventory

def main():
    while True:
        c=int(input("\nEnter a choice:\n1.Add Product\n2.Search Product\n3.Remove Product\n4.view Product\n5.Exit\n\n"))
        if c==1:
            product=input("Enter a product name for add to inventory:").lower()
            add_product(product)
        elif c==2:
            product=input("Search here for product:").lower()
            print(search(product))
        elif c==3:
            product=input("Enter a product name to remove:").lower()
            print(remove(product))
        elif c==4:
            print(display())
        elif c==5:
            print("Bye Bye....!")
            break
        else:
            print("Choose a valid choice\n")
main()
        
            
```
