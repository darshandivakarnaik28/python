### Write a python function find_largest() that accepts the 3 integers as parameters and returns the largest among them

```python
def find_largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>c:
        return b
    else:
        return c

a=input("Enter 1st number:")
b=input("Enter 2st number:")
c=input("Enter 3st number:")
print(find_largest(a,b,c))
        
```

```python
def find_largest(*args):
    largest=float('-inf')
    for i in args:        
        if i>largest:
            largest=i
    return largest
print(find_largest(-120,-10,-10,-20,-30,-40,-100))
```

### Count digits

```python
def count_digits(num):
    count=0
    for i in num:
        i=int(i)
        if i!=0:
            if type((i))==int:
                count+=1
            else:
                return 0
        else:
            return 1    
    
    return count
num=list(input("Enter numbers:"))
print(count_digits(num))
```

```python
def count_digits(a):
    count=0
    if a==0:
        return 1
    while a>0:
        #digit=a%10
        count+=1
        a//=10
    
    return count
num=int(input("Enter numbers:"))
print(count_digits(num))
```

### Sum of digits

```python
def sums(a):
    count=0
    sum=0
    if a==0:
        return 1
    while a>0:
        digit=a%10
        sum+=digit
        count+=1
        a//=10    
    return sum
num=int(input("Enter numbers:"))
print(sums(num))
```

### Palindrome

```python
def Palindrome(a):
    temp=a
    rev=0
    while a>0:
        digit=a%10
        rev=(rev*10)+digit
        a//=10  
    if rev==temp:
        print("Palindrome")
    else:
        print("Not")
num=int(input("Enter numbers:"))
Palindrome(num)
```

### Factorial

```python
def fact(a):
    if a==0 or a==1:
        return 1
    n=1
    while a>1:
        n*=a
        a-=1      
    print(n)
num=int(input("Enter numbers:"))
fact(num)
```

### Prime number

```python
def fact(a):
    n=1
    while a>1:
        n*=a
        a-=1      
    return n
n=int(input("Enter a n:"))

if n>=2:
    if fact(n-1)%n==n-1:
        print("Prime")
    else:
        print("not Prime")
else:
    print( "Prime")

```

### Fibonacci Series of a number

```python
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        n1=0
        n2=1
        i=2
        while i<=n:
            #n3=n1+n2
            #n1=n2
            #n2=n3
            n1,n2=n2,n1+n2
            i+=1
        return n2
n=int(input("enter a number:"))
print(fibonacci(n))

    
    
```
