```python
students=[
    {
        
        "id":101,
        "name":"Alice",
        "course":"Python",
        "attendance":92
    },
    {
        "id":102,
        "name":"Bob",
        "course":"Java",
        "attendance":65
    },
    {
        "id":103,
        "name":"Raj",
        "course":"Java_script",
        "attendance":80
    },
    {
        "id":104,
        "name":"Ram",
        "course":"Unix",
        "attendance":67
    }
]
    
```

### Print all students name

```python
for dct in students:
    print(f"Students are:{dct["name"]}")
```

### students who enrolled for python and java script

```python
course=["Python","Java"]
for dct in students:
    if dct["course"] in course:
        print(dct["name"])
```

### Display name who's attendance is less 75

```python
for dct in students:
    if dct["attendance"]>75:
        print(dct["name"])
```

### Search whether student is present or not

```python
name=input("Enter a name")
for dct in students:
    details=dct["name"]    
    if name in details:
        print("Name found")
        break
else:
    print("Not found")
```

### Reverse a string

```python
name='python'
i=-1
while i>-(len(name)+1):
    print(name[i],end="")
    i-=1
```

### take two variable user name and password

```python
users=[{"name":"darshan","password":"D@123"},{"name":"Darshan","password":"d@123"}]
def verify(username="",password=""):
    for dict in users:
        
        if dict["name"]==username and dict["password"]==password:
            return "Match"
    else:
        return "Dont Match"
b=verify("Darshan","d@123")
print(b)
```
