### check whether a given input is palindrome or not

```python
check="malayalam"
rev=check[::-1]
if check==rev:
    print("palindrome")
else:
    print("not palindrome")
```

```python
check="malayalam"
rev=""
for i in check:
    rev=i+rev
if rev==check:
    print("palindrome")
else:
    print("not palindrome")
```

### check whether a given input is alphabet,number or digit

```python
import re
a=input("Enter a input:")
if len(a)==1:
    if re.match(r"^[a-zA-Z]$",a):
        print("Alphabet")
    elif re.match(r"^[0-9]$",a):
        print("digit")
    else:
        print("symbol")
else:
    print("Too many charters are not allowed")
```

### Write a program to prinnt the color name in VIBGYOR using switch case

```python
color=input("Enter one of these inputs (v,i,b,g,y,o,r):")
match(color.lower()):
    case 'v':print("Violet color")
    case 'i':print("Indigo color")
    case 'b':print("Blue color")
    case 'g':print("Green color")
    case 'y':print("Yello color")
    case 'o':print("Orange color")
    case 'r':print("Red color")
    case _:print("Invalid input")
```
